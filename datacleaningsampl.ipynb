```python
print("Missing Final_Amount:",
      df["Final_Amount"].isnull().sum())

print("\nFinal_Amount summary:")
print(df["Final_Amount"].describe())

print("\nSample values:")
print(df[[
    "Order_Value",
    "Discount_Applied",
    "Final_Amount"
]].head(20))
```

```python
# Calculate Final_Amount where it is missing
# and both Order_Value and Discount_Applied are available

mask = (
    df["Final_Amount"].isna() &
    df["Order_Value"].notna() &
    df["Discount_Applied"].notna()
)

df.loc[mask, "Final_Amount"] = (
    df.loc[mask, "Order_Value"] -
    df.loc[mask, "Discount_Applied"]
)

print("Final_Amount values calculated:", mask.sum())

print("Remaining Final_Amount missing:",
      df["Final_Amount"].isnull().sum())
```

```python
df["Discount_Applied"] = df["Discount_Applied"].fillna(0)

print("Remaining Discount_Applied missing:",
      df["Discount_Applied"].isnull().sum())

df["Final_Amount"] = df["Final_Amount"].fillna(
    df["Order_Value"] - df["Discount_Applied"]
)

print("Remaining Final_Amount missing:",
      df["Final_Amount"].isnull().sum())

```

```python
print("Missing Peak_Hour:",
      df["Peak_Hour"].isnull().sum())

print("\nPeak_Hour values:")
print(df["Peak_Hour"].value_counts(dropna=False))

print("\nData type:")
print(df["Peak_Hour"].dtype)
```

```python
# Fill missing Peak_Hour values with False
df["Peak_Hour"] = df["Peak_Hour"].fillna(False)

# Check the result
print("Missing Peak_Hour after:",
      df["Peak_Hour"].isnull().sum())

print("\nPeak_Hour values after cleaning:")
print(df["Peak_Hour"].value_counts())
```
